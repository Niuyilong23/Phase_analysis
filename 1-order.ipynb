```python
from pymatgen.core import Structure
from pymatgen.transformations.standard_transformations import OrderDisorderedStructureTransformation, SubstitutionTransformation
```

```python
species_map = {"Mn":{"Mn": 0.6, "Li":0.2, "Ni":0.2}}
oxidation_state_dict = {"Li": 1, "Ni": 3, "Mn": 4, "O": -2}
substitute = SubstitutionTransformation(species_map)
structure = Structure.from_file("Mn-Li2MnO3_521.vasp")
```

```python
# 建立无序结构
disordered_structure = substitute.apply_transformation(structure)
# 添加氧化态
disordered_structure.add_oxidation_state_by_element(oxidation_states=oxidation_state_dict)
```

```python
disordered_structure
```

```python
order_calculator =  OrderDisorderedStructureTransformation()
ordered_structure = order_calculator.apply_transformation(disordered_structure, return_ranked_list=100)
```

```python
for i, order_struct in zip(range(len(ordered_structure)), ordered_structure):
    struct_name = "_".join(["file", str(i)]) + ".vasp"
    order_struct["structure"].to(struct_name, fmt="poscar")
```
