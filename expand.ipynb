# 拓展元素掺杂
即，根据已经得到的有序结构进行元素的变换，然后计算其能量。

```python
from pymatgen.core import Structure
import os
```

```python
element_2 = ["Ni", "Co", "Mg", "Zn", "Fe", "Cu"] # 二价元素
element_3 = ["Ni", "Fe", "Sc", "Al", "Ga", "Co"] # 三价元素
element_4 = ["Mn", "Ru", "Sn", "Ti", "Zr", "Ir"] # 四价金属
```

```python
combination_26 = []
for i in element_2:
    for j in element_4:
        combination_26.append([i, j])

combination_44= []
for i in element_3:
    for j in element_4:
        combination_44.append([i, j])
```

```python
ss_benchmark_26 = Structure.from_file("template_file/26-ss.vasp")
ss_benchmark_44 = Structure.from_file("template_file/44-ss.vasp")
segre_benchmark_26 = Structure.from_file("template_file/26-segregation.vasp")
segre_benchmark_44 = Structure.from_file("template_file/44-segregation.vasp")
```

```python
current_dir = os.getcwd()
for com_26 in combination_26:

    work_dirname = "_".join(com_26)
    work_path = os.path.join(current_dir, "26", work_dirname)
    os.mkdir(work_path)

    solid_solution = ss_benchmark_26.copy()
    solid_solution.replace_species({"Ni": com_26[0], "Mn": com_26[1]})
    ss_path = os.path.join(work_path, "ss")
    os.mkdir(ss_path)
    solid_solution_path = os.path.join(ss_path, "POSCAR")
    solid_solution.to(solid_solution_path, fmt="poscar")


    segregation = segre_benchmark_26.copy()
    segregation.replace_species({"Ni": com_26[0], "Mn": com_26[1]})
    segre_path = os.path.join(work_path, "segre")
    os.mkdir(segre_path)
    segregation_path = os.path.join(segre_path, "POSCAR")
    segregation.to(segregation_path, fmt="poscar")




for com_44 in combination_44:
    work_dirname = "_".join(com_44)
    work_path = os.path.join(current_dir, "44", work_dirname)
    os.mkdir(work_path)


    solid_solution = ss_benchmark_44.copy()
    solid_solution.replace_species({"Ni": com_44[0], "Mn": com_44[1]})
    ss_path = os.path.join(work_path, "ss")
    os.mkdir(ss_path)
    solid_solution_path = os.path.join(ss_path, "POSCAR")
    solid_solution.to(solid_solution_path, fmt="poscar")

    segregation = segre_benchmark_44.copy()
    segregation.replace_species({"Ni": com_44[0], "Mn": com_44[1]})
    segre_path = os.path.join(work_path, "segre")
    os.mkdir(segre_path)
    segregation_path = os.path.join(segre_path, "POSCAR")
    segregation.to(segregation_path, fmt="poscar")
```
